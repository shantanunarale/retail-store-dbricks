```python
from pyspark import pipelines as dp
from pyspark.sql.functions import col, expr, lit
```

```python
catalog = spark.conf.get("catalog")
filePath = f"/Volumes/{catalog}/landing/retail_store/orders/"
```

```python
@dp.table(
    name='bronze.orders_raw'
)
def ret_table():
    return spark.readStream \
        .format("cloudFiles") \
            .option("cloudFiles.format", "json") \
            .option("cloudFiles.schemaLocation", "/Volumes/retail_store_dev/landing/retail_store/order_bronze_etl_checkpoint") \
            .option("cloudFiles.schemaHints", "line_items array<struct<product_id int, quantity int>>") \
                .option("cloudFiles.schemaEvolutionMode", "rescue") \
                    .load(filePath)
```
