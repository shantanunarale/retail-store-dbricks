# Setup External Location

```python
%sql
BEGIN
  IF :env == 'dev' THEN
    CREATE EXTERNAL LOCATION IF NOT EXISTS ext_retail_store_dev
    URL 'abfss://retail-store@stexdevcentralindia001.dfs.core.windows.net/'    
    WITH (STORAGE CREDENTIAL ext_storage_dev)
    COMMENT 'External Location for retail store';
    GRANT BROWSE, CREATE EXTERNAL TABLE, CREATE EXTERNAL VOLUME, READ FILES, WRITE FILES ON EXTERNAL LOCATION ext_retail_store_dev TO developers;    
  ELSEIF :env == 'test' THEN
    CREATE EXTERNAL LOCATION IF NOT EXISTS ext_retail_store_stg
    URL 'abfss://retail-store@stexstgcentralindia001.dfs.core.windows.net/'    
    WITH (STORAGE CREDENTIAL ext_storage_stg)
    COMMENT 'External Location for retail store';    
    GRANT BROWSE, READ FILES ON EXTERNAL LOCATION ext_retail_store_stg TO businessusers;
    GRANT BROWSE, READ FILES ON EXTERNAL LOCATION ext_retail_store_stg TO developers;
  END IF;
END;
```
