# Setup External Location

```python
%sql
BEGIN
  IF :env == 'dev' THEN
    CREATE EXTERNAL LOCATION ext_retail_store
    URL 'abfss://retail-store@stexdevcentralindia001.dfs.windows.core.net/'    
    WITH (STORAGE CREDENTIAL ext_storage_dev)
    COMMENT 'External Location for retail store';
  ELSEIF :env == 'stg' THEN
    CREATE EXTERNAL LOCATION ext_retail_store
    URL 'abfss://retail-store@stexstgcentralindia001.dfs.windows.core.net/'    
    WITH (STORAGE CREDENTIAL ext_storage_stg)
    COMMENT 'External Location for retail store';
  END IF;
END;
```
