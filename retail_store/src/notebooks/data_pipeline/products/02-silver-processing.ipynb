```python
from delta.tables import DeltaTable
```

```python
catalog = dbutils.widgets.get("catalog")
```

```python
dt = DeltaTable.forName(spark, f"{catalog}.silver.product")
df = spark.table(f"{catalog}.bronze.product")
dt.alias('t').merge(
    df.alias('s'),
    "s.ProdId = t.ProdId"  
) \
    .whenMatchedUpdateAll(
        condition= "t.Name != s.Name or t.Brand != s.Brand or t.Color != s.Color or t.Category != s.Category"        
    ) \
        .whenNotMatchedInsertAll() \
        .execute()
```
