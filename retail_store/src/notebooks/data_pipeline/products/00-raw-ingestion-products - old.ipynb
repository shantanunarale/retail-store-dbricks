```python
from pyspark.sql.functions import lit, expr, col, current_timestamp
```

```python
catalog = dbutils.widgets.get("catalog")
```

```python
path = f"/Volumes/{catalog}/landing/retail_store/product/"
schema = 'ProdId INT, Name STRING, Brand STRING, Color STRING, Category STRING'
prod_df = spark.read \
    .format("csv") \
        .option("header", "true") \
        .schema(schema) \
            .load(path, pathGlobFilter = '*.csv')     
prod_df = prod_df.withColumn("LoadDate", current_timestamp())              
```

```python
prod_df.write \
    .mode("append") \
        .saveAsTable(f"{catalog}.bronze.product")
```
